```python
#This creates a "canvas" of 20 x 20 
#Every pixel starts at 0 which is black
#Because dtype=uint8, range is 0 - 255 for black to white
img = np.zeros((20, 20), dtype=np.uint8)

#changes the color of the pixels in the range specified to a level of gray 
img[8:15, 10:17] = 50

#change color of specifc block using coordinates
#coordinates are [y,x]
img[3,15] = 200

#modify pixels using a loop 
for x in range(0, 20, 2):
    img[0, x] = 125 #ends up looking like a dotted line

#modify using a mathematical formula
for y in range(20):
    for x in range(20):
        if x == y:
            img[y, x] = 255
plt.axis('off')
plt.imshow(img, cmap='gray', vmin=0, vmax=255)
plt.show()
```
